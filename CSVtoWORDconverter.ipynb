# Script to convert CSV data to WORD for analysis in Atlas.ti

This script was written to help students who collect social media data as CSV tables prepare them for qualitative analysis in Atlas.ti. Unfortunately, Atlas.ti allows only a limited range of input formats, so this scripts adds the metadata and the actual social media posts to a Word document.

To use the script, please press the ```run all``` button in the top bar and wait for the installations below to finish. The script will then ask you to upload your CSV file and will automatically perform the conversion.

**Important**: The CSV file that you upload can have as many columns as you like, bit it **must** have at least the following: ```comment```, ```replies````, ```date```, and ```user_name```. If any of these is missing, the code will break! If you have previous experience with programming, you can, of course, adjust the column names below, but only do it when you are sure that your changes won't result in errors.


`````python
# Install and import required packages

!pip install -q python-docx

import pandas as pd
from docx import Document
from google.colab import files
import io

print("Installation complete!")
`````

`````python
# Upload CSV file
uploaded = files.upload()
csv_file = next(iter(uploaded))
df = pd.read_csv(io.BytesIO(uploaded[csv_file]))

# Create a Word document
doc = Document()
doc.add_heading('CSV to Word Conversion', level=1)

# Add data to the document
for _, row in df.iterrows():
    doc.add_heading(f"User: {row['user_name']} | Date: {row['date']}", level=2)
    doc.add_paragraph(f"Comment: {row['comment']}")
    if not pd.isna(row['replies']):
        doc.add_paragraph(f"Replies: {row['replies']}")
    doc.add_paragraph("---")

# Save the document to a file
doc.save('converted_data.docx')

# Download the Word file
files.download('converted_data.docx')
`````

Once you have successfully downloaded your WORD file, you can rerun the second cell of the script (or the entire script, but it takes longer) to convert additional files.
